#### 4. List comprehension istifadə edərək Selsi ilə verilmiş temperaturlardan yalnız müsbət olanları Farengeytə çevirin. Düstur: F = C × 9/5 + 32 
###### Misal: celsius = [-5, 0, 10, 20, 30, -10, 15]
###### Cavab: [50.0, 68.0, 86.0, 59.0]


```python
celsius = [-5, 0, 10, 20, 30, -10, 15]
f_l=[]
for i in celsius:
    if i >0:
        f=i*9/5+32
        f_l.append(f)
print(f_l)
    
```
